```python
from transformers import AutoModel, AutoTokenizer
```

```python
s = "bert-base-uncased"
token = AutoTokenizer.from_pretrained(s)
```

```python
model = AutoModel.from_pretrained(s)
```

```python
word = "Hello Ai World"
input_ids = token(word, return_tensors="pt")["input_ids"]
print(input_ids)
```

```python
# token to original string
word = token.decode(input_ids[0], skip_special_tokens=True)
print(word)
```

```python
vector_embeddings = model.embeddings.word_embeddings(input_ids)
print(vector_embeddings)
```

```python
import torch

# 1. Measure similarity between your vectors and the entire vocabulary matrix
vocab_weights = model.embeddings.word_embeddings.weight  # shape: [vocab_size, hidden_dim]
similarity = torch.matmul(vector_embeddings, vocab_weights.T)  # shape: [batch_size, seq_len, vocab_size]

# 2. Get the token ID with the highest similarity for each position
predicted_ids = torch.argmax(similarity, dim=-1)

# 3. Decode the token IDs back to text
recovered_text = token.decode(predicted_ids[0], skip_special_tokens=True)
print(recovered_text)  # "i love"
```
